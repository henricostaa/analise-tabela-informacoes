```python
import pandas as pd

df = pd.read_csv('informacoes.csv')

display(df)
```

```python
# tranformar os tipos de dados para o formato correto
# ultimo_login, data_nascimento

df['ultimo_login'] = pd.to_datetime(df['ultimo_login'])
df['data_nascimento'] = pd.to_datetime(df['data_nascimento'])
df = df.drop_duplicates()

display(df.info())
```

```python
# agrupar ultimo_login com cidade
df_grouped = df.groupby('cidade')['ultimo_login'].count().reset_index()
# remover nan
df_grouped = df_grouped.dropna()

df_grouped.sort_values(by='ultimo_login', ascending=False, inplace=True)

display(df_grouped)


```

```python


import plotly.express as px

#mostrar grafico de barras com a quantidade de logins por cidade
fig = px.bar(df_grouped.head(5), x='cidade', y='ultimo_login', title='Quantidade de logins por cidade',)
fig.show()
```

```python
# agrupar ultimo_login com genero
df_grouped = df.groupby('genero')['ultimo_login'].count().reset_index()
# remover nan
df_grouped = df_grouped.dropna()
df_grouped.sort_values(by='ultimo_login', ascending=False, inplace=True)
display(df_grouped)
```

```python
# grafico de barras com a quantidade de logins por genero
fig = px.bar(df_grouped, x='genero', y='ultimo_login', title='Quantidade de logins por genero',)
fig.show()
```

```python
# remover nan de data_nascimento
df = df.dropna(subset=['data_nascimento'])

# grafico de barras com a quantidade de aniversarios por mes
df['mes'] = df['data_nascimento'].dt.month_name()

# ordenar os meses
meses_ordenados = ['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September', 'October', 'November', 'December']
df['mes'] = pd.Categorical(df['mes'], categories=meses_ordenados, ordered=True)

df_grouped = df.groupby('mes').size().reset_index(name='aniversarios')
fig = px.bar(df_grouped, x='mes', y='aniversarios', title='Quantidade de aniversários por mês')
fig.show()
```
